### custom 폐가지수
- 단독주택, 다세대주택만 고려
- min-max scaling
- 0.7* 단독주택 + 0.3 * 다세대주택

```python
df = df[df['행정구역별(시군구)'] != df['결합행정구역']]
df = df[df['주택의 종류별'].isin(['단독주택', '다세대주택'])].reset_index(drop=True)


df.reset_index(drop=True, inplace=True)
df
```

```python
# 연도별 폐가지수를 계산 (예: 2023년 기준)
pivot_df = df.pivot(index='결합행정구역', columns='주택의 종류별', values='2023 년')

# 결측값 처리 (빈집 수가 없으면 0으로 간주)
pivot_df = pivot_df.fillna(0)
```

```python
# 변환 대상 열 지정
columns_to_convert = ['단독주택', '다세대주택']

# 변환 수행
for col in columns_to_convert:
    pivot_df[col] = (
        pivot_df[col]
        .astype(str)                    
        .str.replace(',', '', regex=False)  
        .str.strip()                     
    )
    pivot_df[col] = pd.to_numeric(pivot_df[col], errors='coerce').fillna(0)

pivot_df['폐가지수_2023_raw'] = pivot_df['단독주택'] * 0.6 + pivot_df['다세대주택'] * 0.4
  

from sklearn.preprocessing import MinMaxScaler

scaler = MinMaxScaler()
pivot_df['폐가지수_2023'] = scaler.fit_transform(pivot_df[['폐가지수_2023_raw']])
```

```python
pivot_df.to_csv('./ref_dataset/폐가지수.csv')
```
